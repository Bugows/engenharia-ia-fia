## Few-shot: os exemplos viram conversa dentro do prompt

```python
from assistente_ia.container import criar_llm
from assistente_ia.domain.modelos import Mensagem, sistema, usuario

exemplos = [json.loads(l) for l in linhas[:3]]
few_shot = [Mensagem(m["role"], m["content"]) for e in exemplos for m in e["messages"][1:]]
persona = exemplos[0]["messages"][0]["content"]
print(criar_llm().responder([sistema(persona), *few_shot, usuario("Vocês fazem bolo de aniversário?")]).texto)
```
