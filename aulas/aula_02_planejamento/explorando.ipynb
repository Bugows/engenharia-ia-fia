## Simulador: mexa nos números e rode de novo

```python
conversas_por_mes = 3000
tokens_entrada = 2500     # prompt de sistema + trechos do RAG + pergunta
tokens_saida = 300
minutos_economizados = 3
custo_hora_atendente = 25.0
custo_fixo = 150.0

custo_tokens = custo_brl(custo_usd("gpt-5.4-mini", tokens_entrada, tokens_saida) * conversas_por_mes)
custo_total = custo_tokens + custo_fixo
ganho = conversas_por_mes * minutos_economizados / 60 * custo_hora_atendente
print(f"custo mensal: {formatar_brl(custo_total)} | ganho: {formatar_brl(ganho)} | ROI: {roi(ganho, custo_total):.0%}")
```

## Pergunta para a turma
Quantos minutos por conversa a IA precisa economizar para o ROI ficar positivo? Encontre com a célula abaixo.

```python
for minutos in [0.05, 0.1, 0.2, 0.5, 1]:
    ganho = conversas_por_mes * minutos / 60 * custo_hora_atendente
    print(f"{minutos:>5} min -> ROI {roi(ganho, custo_total):>6.0%}")
```
