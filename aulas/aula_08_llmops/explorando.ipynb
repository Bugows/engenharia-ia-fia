```python
from assistente_ia.application.chat import ServicoChat
from assistente_ia.container import criar_llm, criar_observabilidade

obs = criar_observabilidade()
chat = ServicoChat(criar_llm(), observabilidade=obs)
for p in ["Oi!", "Tem pão de queijo?", "E café?"]:
    print(">", chat.perguntar(p))
obs.resumo()
```

```python
import time
for pedaco in criar_llm().transmitir([*chat.historico[-2:]]):
    print(pedaco, end="", flush=True)
    time.sleep(0.03)
```
