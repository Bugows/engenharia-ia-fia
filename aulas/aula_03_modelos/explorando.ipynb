## Embeddings: escreva suas palavras e veja o mapa de calor

```python
%matplotlib inline
import matplotlib.pyplot as plt
from assistente_ia.adapters.vector_store.memoria import cosseno
from assistente_ia.container import criar_embeddings

palavras = ["aipim", "mandioca", "macaxeira", "semáforo", "sinaleira", "farol", "futebol"]
vetores = criar_embeddings().vetorizar_textos(palavras)
matriz = [[cosseno(a, b) for b in vetores] for a in vetores]

plt.figure(figsize=(7, 5))
plt.imshow(matriz, cmap="YlOrBr", vmin=0, vmax=1)
plt.xticks(range(len(palavras)), palavras, rotation=45, ha="right")
plt.yticks(range(len(palavras)), palavras)
plt.colorbar(); plt.title("Similaridade de cosseno"); plt.tight_layout()
```

## Busca semântica em 4 linhas
Qual frase do cardápio é a mais parecida com a pergunta?

```python
frases = ["Pão de queijo quentinho às 7h", "Café coado na hora", "Bolo de rolo pernambucano", "Entrega em até 3 km"]
pergunta = "Vocês levam em casa?"
emb = criar_embeddings()
vetores = emb.vetorizar_textos(frases)
consulta = emb.vetorizar_consulta(pergunta)
sorted(((round(cosseno(consulta, v), 3), f) for v, f in zip(vetores, frases)), reverse=True)
```

> No modo fake, 'levam em casa' não encontra 'Entrega'. Com o embedding real, encontra. Essa é a diferença entre comparar **palavras** e comparar **significado**.
